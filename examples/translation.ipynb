# Translating an Ubermag system

This example checks the in-memory TOMMOS representation produced by `tommos-adapter`.

```python
import micromagneticmodel as mm
import numpy as np
import pyvista as pv
import tommos_adapter as ta

points = np.array([[0, 0, 0], [1e-9, 0, 0], [0, 1e-9, 0], [0, 0, 1e-9]])
grid = pv.UnstructuredGrid(
    np.array([4, 0, 1, 2, 3]),
    np.array([pv.CellType.TETRA], dtype=np.uint8),
    points,
)
grid.point_data["m"] = np.tile((0.0, 0.0, 1.0), (4, 1))
grid.cell_data["Ms"] = np.array([1e6])
system = mm.System(
    name="tetrahedron",
    energy=mm.Exchange(A=1e-11) + mm.Demag() + mm.Zeeman(H=(0, 0, 1e6)),
    m=grid,
)

mesh = ta.scripts.mesh_data(system)
assert mesh["knt"].shape == (4, 3)
assert mesh["ijk"].shape == (1, 5)
assert len(ta.scripts.krn_script(system).splitlines()) == 1
assert "[mesh]" in ta.scripts.p2_script(system)
```
